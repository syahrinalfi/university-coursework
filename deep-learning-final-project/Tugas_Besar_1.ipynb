## Bagi Data

```python
# Define dataset directory
dataset_dir = "TrashType_Image_Dataset"
train_dir = "dataset/train"
val_dir = "dataset/val"
test_dir = "dataset/test"
```

```python
# Create train and val directories
os.makedirs(train_dir, exist_ok=True)
os.makedirs(val_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)
```

```python
# Get list of object folders
object_folders = [f for f in os.listdir(dataset_dir) if os.path.isdir(os.path.join(dataset_dir, f))]
```

```python
# Split data
for obj_folder in object_folders:
    # Paths for object folder
    obj_path = os.path.join(dataset_dir, obj_folder)
    obj_train_path = os.path.join(train_dir, obj_folder)
    obj_val_path = os.path.join(val_dir, obj_folder)
    obj_test_path = os.path.join(test_dir, obj_folder)

    # Create directories for train and val data
    os.makedirs(obj_train_path, exist_ok=True)
    os.makedirs(obj_val_path, exist_ok=True)
    os.makedirs(obj_test_path, exist_ok=True)

    # List of images in the object folder
    images = [img for img in os.listdir(obj_path) if img.lower().endswith(('.png', '.jpg', '.jpeg'))]

    # Split into train and val
    train_images, val_images = train_test_split(images, test_size=0.3, random_state=42, shuffle=True)
    # Split into train and val
    val_images, test_images = train_test_split(val_images, test_size=0.5, random_state=42, shuffle=True)

    # Move files to train and val directories
    for img in train_images:
        sh.copy2(os.path.join(obj_path, img), os.path.join(obj_train_path, img))
    for img in val_images:
        sh.copy2(os.path.join(obj_path, img), os.path.join(obj_val_path, img))
    for img in test_images:
        sh.copy2(os.path.join(obj_path, img), os.path.join(obj_test_path, img))

print(f"Data has been split and organized into:/nTrain directory: {train_dir}/nval directory: {val_dir}")
```

## Training

```python
model = YOLO("yolo11n-cls.pt")
model.train(data="dataset", epochs=20)
```
